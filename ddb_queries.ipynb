```python
import duckdb

ddb = '../loan_data.duckdb'
```

```python
with duckdb.connect(ddb) as conn:
    print(conn.sql('SELECT * FROM duckdb_tables()'))
```

```python
with duckdb.connect(ddb) as conn:
    print(conn.sql("""
                   SELECT int_rate, id
                   FROM raw_loan_data
                   WHERE SUBSTRING(issue_d, 5, 8) = '2020'
                   order by id
                   """))
```

```python
with duckdb.connect(ddb) as conn:
    print(conn.sql("""
                   SELECT interest_rate, loan_id, purpose_description
                   FROM stg_loan_data_selected_cols
                   WHERE DATE_PART('year',issue_date) = '2020'
                   ORDER BY loan_id
                   """))
```

```python
with duckdb.connect(ddb) as conn:
    print(conn.sql("""
                   SELECT 
                   COUNT(*)
                   FROM raw_loan_data
                   WHERE SUBSTRING(issue_d, 5, 8) >= '2018'
                   ORDER BY COUNT(*)
                   """))
```

```python
with duckdb.connect(ddb) as conn:
    print(conn.sql('DESCRIBE stg_loan_data_selected_cols'))
```
